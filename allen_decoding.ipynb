```python
session_id = selected_session_ids[0]

session = cache.get_session_data(session_id)

print("Loaded session:", session_id)
print(type(session))
```

```python
units = session.units.copy()

print(units.shape)
print(units.columns.tolist())

display(
    units[
        [
            "ecephys_structure_acronym",
            "firing_rate",
            "presence_ratio",
            "amplitude_cutoff",
            "isi_violations",
        ]
    ].describe(include="all")
)
# filter bad neurons
target_areas = ["VISp", "VISl", "VISpm"]

good_units = units[
    units["ecephys_structure_acronym"].isin(target_areas)
    & (units["presence_ratio"] > 0.9)
    & (units["amplitude_cutoff"] < 0.1)
    & (units["isi_violations"] < 0.5)
].copy()

print("Units before filtering:")
print(
    units[
        units["ecephys_structure_acronym"].isin(target_areas)
    ]["ecephys_structure_acronym"].value_counts()
)

print("\nUnits after filtering:")
print(
    good_units[
        "ecephys_structure_acronym"
    ].value_counts()
)
```

```python
print(
    binary_trials["spatial_frequency"]
    .value_counts(dropna=False)
)
simple_trials = binary_trials[
    binary_trials["spatial_frequency"] == 0.02
].copy()

print(
    simple_trials["orientation"]
    .value_counts()
)
```

```python
#print(simple_trials["phase shift"].value_counts(dropna=False))
```

```python
y = (
    simple_trials["orientation"] == 90.0
).astype(int).to_numpy()

print(
    pd.Series(y).value_counts().sort_index()
)
```

```python
unit_ids = good_units.index.to_numpy()

print("Number of neurons:", len(unit_ids))
```

```python
spike_times = session.spike_times

print(type(spike_times))
print("Number of units with spike times:", len(spike_times))
```

```python
missing_unit_ids = [
    unit_id
    for unit_id in unit_ids
    if unit_id not in spike_times
]

print("Missing selected neurons:", len(missing_unit_ids))
```

```python
trial_onsets = simple_trials["start_time"].to_numpy()

print("Number of trial onsets:", len(trial_onsets))
print("First 5 onset times:", trial_onsets[:5])

print(
    simple_trials["duration"]
    .describe()
)
```

```python
window_start = 0.0
window_end = 0.25

print("Response window:", window_start, "to", window_end, "seconds")
```

```python
X = np.zeros(
    (len(simple_trials), len(unit_ids)),
    dtype=np.float32,
)

print("X shape:", X.shape)
print("Number of unit IDs:", len(unit_ids))
print("Number of rows in good_units:", len(good_units))
print("Number of columns in X:", X.shape[1])
```

```python
for neuron_index, unit_id in enumerate(unit_ids):
    spikes = np.asarray(spike_times[unit_id])

    start_indices = np.searchsorted(
        spikes,
        trial_onsets + window_start,
        side="left",
    )

    end_indices = np.searchsorted(
        spikes,
        trial_onsets + window_end,
        side="left",
    )

    X[:, neuron_index] = (
        end_indices - start_indices
    )
```

```python
print("X shape:", X.shape)
print("X dtype:", X.dtype)
print("Minimum spike count:", X.min())
print("Maximum spike count:", X.max())
print("Mean spike count:", X.mean())
print("Fraction of zero entries:", np.mean(X == 0))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42,
)

print("Training shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("Training labels:", np.bincount(y_train))
print("Test labels:", np.bincount(y_test))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled test shape:", X_test_scaled.shape)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

cv_model = Pipeline([
    ("scaler", StandardScaler()),
    (
        "classifier",
        LogisticRegression(
            max_iter=2000,
            random_state=42,
        ),
    ),
])

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42,
)

cv_scores = cross_val_score(
    cv_model,
    X,
    y,
    cv=cv,
    scoring="accuracy",
)

print("Fold accuracies:", cv_scores)
print("Mean accuracy:", cv_scores.mean())
print("Standard deviation:", cv_scores.std())
```

```python
area_scores = {}

for area in ["VISp", "VISl", "VISpm"]:
    area_mask = unit_areas == area
    X_area = X[:, area_mask]

    scores = cross_val_score(
        cv_model,
        X_area,
        y,
        cv=cv,
        scoring="accuracy",
    )

    area_scores[area] = scores

    print(
        area,
        "neurons:", X_area.shape[1],
        "mean accuracy:", scores.mean(),
        "std:", scores.std(),
    )
```

```python
n_neurons_equal = 40
n_repeats = 20

rng = np.random.default_rng(42)

print("Neurons per area:", n_neurons_equal)
print("Number of repeated samples:", n_repeats)

equal_size_results = {
    "VISp": [],
    "VISl": [],
    "VISpm": [],
}

for area in ["VISp", "VISl", "VISpm"]:
    area_indices = np.where(unit_areas == area)[0]

    for _ in range(n_repeats):
        sampled_indices = rng.choice(
            area_indices,
            size=n_neurons_equal,
            replace=False,
        )

        X_sampled = X[:, sampled_indices]

        scores = cross_val_score(
            cv_model,
            X_sampled,
            y,
            cv=cv,
            scoring="accuracy",
        )

        equal_size_results[area].append(
            scores.mean()
        )

for area in ["VISp", "VISl", "VISpm"]:
    values = np.array(equal_size_results[area])

    print(
        area,
        "mean accuracy:", values.mean(),
        "std across samples:", values.std(),
        "min:", values.min(),
        "max:", values.max(),
    )
```

```python
six_orientation_trials = static_trials[
    (static_trials["spatial_frequency"] == 0.02)
    & (static_trials["orientation"] != "null")
].copy()

print(
    six_orientation_trials["orientation"]
    .value_counts()
    .sort_index()
)
```

```python
six_y = six_orientation_trials["orientation"].astype(float).to_numpy()
six_trial_onsets = six_orientation_trials["start_time"].to_numpy()

print("Number of trials:", len(six_y))
print("Labels:", np.unique(six_y))
```

```python
six_X = np.zeros(
    (len(six_orientation_trials), len(unit_ids)),
    dtype=np.float32,
)

for neuron_index, unit_id in enumerate(unit_ids):
    spikes = np.asarray(spike_times[unit_id])

    start_indices = np.searchsorted(
        spikes,
        six_trial_onsets + window_start,
        side="left",
    )

    end_indices = np.searchsorted(
        spikes,
        six_trial_onsets + window_end,
        side="left",
    )

    six_X[:, neuron_index] = end_indices - start_indices

print("Feature matrix shape:", six_X.shape)
print("Mean spike count:", six_X.mean())
print("Maximum spike count:", six_X.max())
```

```python
six_cv_scores = cross_val_score(
    cv_model,
    six_X,
    six_y,
    cv=cv,
    scoring="accuracy",
)

print("Fold accuracies:", six_cv_scores)
print("Mean accuracy:", six_cv_scores.mean())
print("Standard deviation:", six_cv_scores.std())
print("Chance level:", 1 / 6)
```

```python
six_area_scores = {}

for area in ["VISp", "VISl", "VISpm"]:
    area_mask = unit_areas == area
    six_X_area = six_X[:, area_mask]

    scores = cross_val_score(
        cv_model,
        six_X_area,
        six_y,
        cv=cv,
        scoring="accuracy",
    )

    six_area_scores[area] = scores

    print(
        area,
        "neurons:", six_X_area.shape[1],
        "mean accuracy:", scores.mean(),
        "std:", scores.std(),
    )
```

```python
six_equal_results = {
    "VISp": [],
    "VISl": [],
    "VISpm": [],
}

rng = np.random.default_rng(42)

for area in ["VISp", "VISl", "VISpm"]:
    area_indices = np.where(unit_areas == area)[0]

    for _ in range(n_repeats):
        sampled_indices = rng.choice(
            area_indices,
            size=40,
            replace=False,
        )

        scores = cross_val_score(
            cv_model,
            six_X[:, sampled_indices],
            six_y,
            cv=cv,
            scoring="accuracy",
        )

        six_equal_results[area].append(scores.mean())

for area in ["VISp", "VISl", "VISpm"]:
    values = np.array(six_equal_results[area])

    print(
        area,
        "mean accuracy:", values.mean(),
        "std across samples:", values.std(),
    )
```
